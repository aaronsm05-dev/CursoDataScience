```python
#Funciones

def clasificar_edad(edad):
    """Recibe edad y regresa la categoria correspondiente"""
    if edad < 12:
        return "Niño"
    elif edad < 18:
        return "Adolescente"
    elif edad < 65:
        return "Adulto"
    else:
        return "Adulto mayor"

#Probamos la funcion con distintos valores
for edad in [5, 15, 40, 80]:
    print(edad, "->", clasificar_edad(edad))
```

```python
#ejercicio

lista_pasajeros = []

def ingresar_pasajero(nombre, edad, clase):
    pasajero = {
        "nombre": nombre,
        "edad": edad,
        "clase": clase
    }
    lista_pasajeros.append(pasajero)

while True:
    if len(lista_pasajeros) <= 4:
        if len(lista_pasajeros) == 0:
            print("La lista esta vacia")
        else:
            print("La lista tiene ", len(lista_pasajeros), "pasajeros")
        respuesta = input("Desea ingresar un pasajero? y/n")
        if respuesta == "y":
            nombre = input("Ingresa el nombre: ")
            edad = input("Ingresa la edad: ")
            clase = input("Ingresa la clase: ")
            ingresar_pasajero(nombre, edad, clase)

            print("Lista de pasajeros: ")
            print(lista_pasajeros)
            
        else:
            break
    else:
        break
               
```
