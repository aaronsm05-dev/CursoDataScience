```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder

sns.set_theme(style="whitegrid")
df = pd.read_csv("titanic.csv")
print("Filas y columnas originales: ", df.shape)
df.head()
```

# Se necesita preparar lo datos poque machine learning solamente entiende nuemros, tambien sensible a Nan (Not a number), texto sin convertir, columnas irrelevantes y escalas muy distintas

```python
#Tratamiento de datos nulos
df.isnull().sum().sort_values(ascending=False)
```

```python
#Copiamos el dataframe oara no modificar el original
df_prep = df.copy()

#1) Cabin: se elimina mas adelante

#2) Age: imputacion con la mediana
mediana_edad = df_prep["Age"].median()
df_prep["Age"] = df_prep["Age"].fillna(mediana_edad)
print("Mediana de edad usada para imputar: ", mediana_edad)

#3) Embarked: Imputacion por la moda
moda_embarque = df_prep["Embarked"].mode()[0]
df_prep["Embarked"] = df_prep["Embarked"].fillna(moda_embarque)
print("Moda de puerto de embarque usada para imputar: ", moda_embarque)

print()
print("Valores nulos despues de la imputacion: ")
print(df_prep.isnull().sum().sort_values(ascending=False))
```

```python
#Borrar columnas con texto e irrelevantes
columnas_a_eliminar = ["PassengerId", "Name", "Ticket", "Cabin"]
df_prep = df_prep.drop(columns= columnas_a_eliminar)
#metodo para eliminar es el drop
print("Columnas restantes: ", df_prep.columns.to_list())
df_prep.head()
```

## Ingenieria de caracteristicas simple

creamos nuevas columnas a partir de las existentes para capturar informacion util que no esta explicita

Ejemplo: El tamaño de la familia a bordo

```python
#Tamaño de la familia a bordo
df_prep["FamilySize"] = df_prep["SibSp"] + df_prep["Parch"] + 1

#Viajaba solo?
df_prep["IsAlone"] = (df_prep["FamilySize"] == 1).astype(int)

df_prep[["SibSp", "Parch", "FamilySize", "IsAlone"]].head(10)
```

```python
sns.barplot(data= df_prep , x= "IsAlone", y= "Survived")
plt.title("Supervivencia segun si viajaba solo (1) o no (0)")
plt.show()
```

## Codificacion de vaiables categoricas

```python
df_prep["Sex"] = df_prep["Sex"].map({"male": 0, "female": 1})
df_prep[["Sex"]].head()
```

```python
df_prep = pd.get_dummies(df_prep, columns=["Embarked"], prefix="Embarked", drop_first=True)
df_prep.head()
```

```python
columnas_numericas = ["Age", "Fare", "FamilySize"]

scaler = StandardScaler()
df_prep[columnas_numericas] = scaler.fit_transform(df_prep[columnas_numericas])

df_prep[columnas_numericas].describe().round(2)
```

# Division de conjunto de entrenamiento y prueba

Usaremos 80% para entrenamiento (train) y 20% para pruebas (test)

```python
x = df_prep.drop(columns=["Survived"])
y = df_prep["Survived"]

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify= y
)

print("Tamaño de xtrain: ", x_train.shape)
print("Tamaño de xtest: ", x_test.shape)
print()
print("Proporcion de sobrevivientes en ytrain:" , y_train.mean().round(3))
print("Proporcion de sobrevivientes en ytest:" , y_test.mean().round(3))
```

```python
x_train.to_csv("X_train.cvs", index= False)
x_test.to_csv("X_test.cvs", index= False)
y_train.to_csv("y_train.cvs", index= False)
y_test.to_csv("y_test.cvs", index= False)
```

```python
def preparar_datos(df_original):
    df_local = df_original.copy()

    #imputar valores nulos (Age y embarked)
    mediana_edad = df_local["Age"].median()
    df_local["Age"] = df_local["Age"].fillna(mediana_edad)
    moda_embarque = df_local["Embarked"].mode()[0]
    df_local["Embarked"] = df_local["Embarked"].fillna(moda_embarque)

    #Eliminar columnas irrelevantes
    columnas_a_eliminar = ["PassengerId", "Name", "Ticket", "Cabin"]
    df_local = df_local.drop(columns= columnas_a_eliminar)

    #Crear FamilySize e isAlone
        #Tamaño de la familia a bordo
    df_local["FamilySize"] = df_local["SibSp"] + df_local["Parch"] + 1
        #Viajaba solo?
    df_local["IsAlone"] = (df_local["FamilySize"] == 1).astype(int)
    df_local[["SibSp", "Parch", "FamilySize", "IsAlone"]].head(10)

    #Codificar sex(0/1) y Embarked(One-hot)
    df_local["Sex"] = df_local["Sex"].map({"male": 0, "female": 1})
    df_local = pd.get_dummies(df_local, columns=["Embarked"], prefix="Embarked", drop_first=True)

    #Escalar age, fare, FamilySize
    columnas_numericas = ["Age", "Fare", "FamilySize"]
    scaler = StandardScaler()
    df_local[columnas_numericas] = scaler.fit_transform(df_local[columnas_numericas])

    #Separar x y y, y dividir en train test
    x = df_local.drop(columns=["Survived"])
    y = df_local["Survived"]
    x_train, x_test, y_train, y_test = train_test_split(
        x, y,
        test_size=0.2,
        random_state=42,
        stratify= y
    )

    return x_train, x_test, y_train, y_test

    
```

```python
X_train2, X_test2, y_train2, ytest2 = preparar_datos(df)
print(X_train2.shape, X_test2.shape)
```
