```python
#importar laslibrerias que vamos a ocupar

import pandas as pd #carga de datos
import numpy  as np #analisis de datos
import matplotlib.pyplot as plt #Gráficas
```

```python
#cargar el archivo de datos (dataframe
df = pd.read_csv('titanic.csv')
print("carga de datos exitosa!!")
```

```python
#inspeccion inicial
df.head() #carga las primeras 5 lineas
```

```python
df.info() #tipos de datos que tenemos
```

```python
df.describe() #estadísticas 
```

```python
df.describe(include= 'all') #estadísticas incluye todas las columnas 
```

Limpieza basica de datos

Tratar con los valores faltantes: Rellenar las columnas con valores faltantes con 
Dropear: Cabin embarked
Convertir: Asegurar que la columna Pclass sea categorico


```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df.head
```

```python
df.dropna(subset=['Embarked'], inplace= True)
```

```python
df.drop('Cabin', axis= 1, inplace= True)
```

```python
df['Pclass'] = df['Pclass'].astype('category')
```

```python
df.info()
```

```python
#Configuracion estética

import seaborn as sns
sns.set(style= "whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
#Variable objetivo
sns.countplot(x= "Survived", data=df)
plt.title("Distribucion de sobrevivientes")
plt.show()
```

```python
#Variables categorias
cat_vars = ["Sex", "Pclass", "Embarked"]
```

```python
for col in cat_vars:
    sns.countplot(x=col, data=df)
    plt.title("Distribucion de {col}")
    plt.show()
```

```python
#variables numericas
num_vars = ["Age", "Fare"]
```

```python
for col in num_vars:
    sns.histplot(df[col].dropna(), kde= True, bins=30)
    plt.title(f"Distribucion de {col}")
    plt.show()
```

```python
#supervivencia por sexo
sns.barplot(x= "Sex", y= "Survived", data=df)
plt.title("Sobrevivientes segun sexo")
plt.show()
```

```python
#supervivencia por clase
sns.barplot(x= "Pclass", y= "Survived", data=df)
plt.title("Sobrevivientes según clase")
plt.show()
```

```python
#Supervivencia por edad
sns.barplot(x="Survived", y="Age", data=df)
plt.title("Sobrevivientes según edad")
plt.show()
```

```python
#Correlaciones
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Mapa de coreelaciones")
plt.show()
```

```python
#Guardar el nuevo dataset limpio
df.to_csv("datos_nuevos_titanic.csv", index=False)
```
